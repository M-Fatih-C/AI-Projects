```python
from ultralytics import YOLO

# Hafif model seçimi (yolov8n.pt)
model = YOLO('yolov8n.pt')

# Eğitim komutu
model.train(
    data='/content/data/data.yaml',   # data.yaml yolunu doğru ver
    epochs=50,                        # 50 epoch önerilir
    imgsz=640,                        # Görsel boyutu
    batch=8,                          # Belleğe göre değiştirilebilir
    project='PlakaEgitimi',          # Eğitim çıktı klasörü
    name='plaka_modeli',             # Eğitim adı
    exist_ok=True                    # Aynı isim varsa üzerine yaz
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/PlakaEgitimi/plaka_modeli/results.csv")

# Epoch bazlı grafikler
plt.figure(figsize=(12, 6))
plt.plot(df['epoch'], df['metrics/precision(B)'], label='Precision')
plt.plot(df['epoch'], df['metrics/recall(B)'], label='Recall')
plt.xlabel("Epoch")
plt.ylabel("Skor")
plt.title("Model Performansı")
plt.legend()
plt.grid()
plt.show()
```
