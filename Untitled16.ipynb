```python
# ═══════════════════════════════════════════════════════════════
# 4. METRİK FONKSİYONLARI
# ═══════════════════════════════════════════════════════════════

def recall_at_k(y_true, y_prob, k=0.1):
    """Tahmin edilen olasılıkların en üst k%'sında recall"""
    y_true = np.asarray(y_true)
    y_prob = np.asarray(y_prob)
    n = len(y_true)
    m = max(1, int(np.round(k * n)))
    order = np.argsort(-y_prob, kind="mergesort")
    top = order[:m]
    tp_at_k = y_true[top].sum()
    P = y_true.sum()
    return float(tp_at_k / P) if P > 0 else 0.0

def lift_at_k(y_true, y_prob, k=0.1):
    """Tahmin edilen olasılıkların en üst k%'sında lift"""
    y_true = np.asarray(y_true)
    y_prob = np.asarray(y_prob)
    n = len(y_true)
    m = max(1, int(np.round(k * n)))
    order = np.argsort(-y_prob, kind="mergesort")
    top = order[:m]
    tp_at_k = y_true[top].sum()
    precision_at_k = tp_at_k / m
    prevalence = y_true.mean()
    return float(precision_at_k / prevalence) if prevalence > 0 else 0.0

def convert_auc_to_gini(auc):
    """AUC'yi Gini'ye çevir"""
    return 2 * auc - 1

def ing_hubs_datathon_metric(y_true, y_prob):
    """Resmi bileşik metrik"""
    score_weights = {
        "gini": 0.4,
        "recall_at_10perc": 0.3,
        "lift_at_10perc": 0.3,
    }
    baseline_scores = {
        "roc_auc": 0.6925726757936908,
        "recall_at_10perc": 0.18469015795868773,
        "lift_at_10perc": 1.847159286784029,
    }

    roc_auc = roc_auc_score(y_true, y_prob)
    recall_at_10perc = recall_at_k(y_true, y_prob, k=0.1)
    lift_at_10perc = lift_at_k(y_true, y_prob, k=0.1)

    new_scores = {
        "roc_auc": roc_auc,
        "recall_at_10perc": recall_at_10perc,
        "lift_at_10perc": lift_at_10perc,
    }

    baseline_scores["gini"] = convert_auc_to_gini(baseline_scores["roc_auc"])
    new_scores["gini"] = convert_auc_to_gini(new_scores["roc_auc"])

    final_gini_score = new_scores["gini"] / baseline_scores["gini"]
    final_recall_score = new_scores["recall_at_10perc"] / baseline_scores["recall_at_10perc"]
    final_lift_score = new_scores["lift_at_10perc"] / baseline_scores["lift_at_10perc"]

    final_score = (
        final_gini_score * score_weights["gini"] +
        final_recall_score * score_weights["recall_at_10perc"] +
        final_lift_score * score_weights["lift_at_10perc"]
    )
    return final_score

def calculate_all_metrics(y_true, y_prob):
    """Tüm metrikleri hesapla"""
    auc = roc_auc_score(y_true, y_prob)
    gini = convert_auc_to_gini(auc)
    recall10 = recall_at_k(y_true, y_prob, k=0.1)
    lift10 = lift_at_k(y_true, y_prob, k=0.1)
    composite = ing_hubs_datathon_metric(y_true, y_prob)

    return {
        "auc": auc,
        "gini": gini,
        "recall@10": recall10,
        "lift@10": lift10,
        "composite_score": composite
    }

def print_metrics(metrics, prefix=""):
    """Metrikleri yazdır"""
    print(f"\n{'='*60}")
    print(f"{prefix} METRİKLER")
    print(f"{'='*60}")
    print(f"AUC:              {metrics['auc']:.5f}")
    print(f"Gini:             {metrics['gini']:.5f}")
    print(f"Recall@10%:       {metrics['recall@10']:.5f}")
    print(f"Lift@10%:         {metrics['lift@10']:.5f}")
    print(f"─────────────────────────────────────────")
    print(f"📊 Composite Score: {metrics['composite_score']:.5f}")
    print(f"{'='*60}\n")

print("✅ Metrik fonksiyonları hazır!")
```

```python
# ═══════════════════════════════════════════════════════════════
# 6. FEATURE ENGINEERING - PART 1 (Yardımcı Fonksiyonlar)
# ═══════════════════════════════════════════════════════════════

class FeatureEngineer:
    """Sızıntısız özellik üretimi"""

    def __init__(self, customer_history, customers):
        self.history = customer_history.copy()
        self.customers = customers.copy()
        self.history['date'] = pd.to_datetime(self.history['date'])

        self.txn_cols = [
            'mobile_eft_all_cnt', 'mobile_eft_all_amt',
            'cc_transaction_all_cnt', 'cc_transaction_all_amt'
        ]

    def create_features(self, reference_df):
        """Ana özellik üretim fonksiyonu"""
        print(f"\n🔧 FEATURE ENGINEERING BAŞLIYOR...")
        print(f"   Hedef: {len(reference_df):,} satır")

        ref_df = reference_df.copy()
        ref_df['ref_date'] = pd.to_datetime(ref_df['ref_date'])

        features_list = []
        total = len(ref_df)

        for idx, row in ref_df.iterrows():
            if idx % 5000 == 0:
                print(f"   İşlenen: {idx:,}/{total:,} ({100*idx/total:.1f}%)")

            cust_id = row['cust_id']
            ref_date = row['ref_date']

            # SIZMA ÖNLEMe: sadece ref_date'ten ÖNCE
            cust_history = self.history[
                (self.history['cust_id'] == cust_id) &
                (self.history['date'] < ref_date)
            ].copy()

            features = {'cust_id': cust_id}

            # 1. Pencereli istatistikler
            features.update(self._windowed_stats(cust_history, ref_date))

            # 2. RFM
            features.update(self._rfm_features(cust_history, ref_date))

            # 3. Trend
            features.update(self._trend_features(cust_history, ref_date))

            # 4. Ürün
            features.update(self._product_features(cust_history, ref_date))

            # 5. Oranlar
            features.update(self._ratio_features(cust_history, ref_date))

            # 6. Zaman
            features.update(self._temporal_features(ref_date))

            # 7. Demografi
            features.update(self._demographic_features(cust_id))

            features_list.append(features)

        features_df = pd.DataFrame(features_list)

        print(f"\n✅ FEATURE ENGINEERING TAMAMLANDI!")
        print(f"   Özellik sayısı: {features_df.shape[1] - 1}")
        print(f"   NaN kontrolü: {features_df.isnull().sum().sum()} NaN bulundu")

        # NaN'ları 0 ile doldur
        features_df = features_df.fillna(0)

        return features_df

    def _windowed_stats(self, cust_history, ref_date):
        """Pencereli istatistikler"""
        features = {}

        for window in config.WINDOWS:
            window_start = ref_date - pd.DateOffset(months=window)
            window_data = cust_history[cust_history['date'] >= window_start]
            suffix = f"{window}m"

            for col in self.txn_cols:
                if len(window_data) > 0:
                    features[f"{col}_sum_{suffix}"] = window_data[col].sum()
                    features[f"{col}_mean_{suffix}"] = window_data[col].mean()
                    features[f"{col}_max_{suffix}"] = window_data[col].max()
                    features[f"{col}_std_{suffix}"] = window_data[col].std() if len(window_data) > 1 else 0
                    features[f"{col}_min_{suffix}"] = window_data[col].min()
                else:
                    features[f"{col}_sum_{suffix}"] = 0
                    features[f"{col}_mean_{suffix}"] = 0
                    features[f"{col}_max_{suffix}"] = 0
                    features[f"{col}_std_{suffix}"] = 0
                    features[f"{col}_min_{suffix}"] = 0

            features[f"active_months_{suffix}"] = len(window_data)
            features[f"is_missing_{suffix}"] = int(len(window_data) == 0)

        return features

    def _rfm_features(self, cust_history, ref_date):
        """RFM özellikleri"""
        features = {}

        if len(cust_history) > 0:
            # Recency
            last_date = cust_history['date'].max()
            features['months_since_last_txn'] = (ref_date.year - last_date.year) * 12 + (ref_date.month - last_date.month)

            # CC/EFT recency
            cc_data = cust_history[cust_history['cc_transaction_all_cnt'] > 0]
            eft_data = cust_history[cust_history['mobile_eft_all_cnt'] > 0]

            if len(cc_data) > 0:
                last_cc = cc_data['date'].max()
                features['months_since_last_cc'] = (ref_date.year - last_cc.year) * 12 + (ref_date.month - last_cc.month)
            else:
                features['months_since_last_cc'] = 999

            if len(eft_data) > 0:
                last_eft = eft_data['date'].max()
                features['months_since_last_eft'] = (ref_date.year - last_eft.year) * 12 + (ref_date.month - last_eft.month)
            else:
                features['months_since_last_eft'] = 999

            # Monetary
            for window in [3, 6, 12]:
                window_start = ref_date - pd.DateOffset(months=window)
                window_data = cust_history[cust_history['date'] >= window_start]

                if len(window_data) > 0:
                    cc_amt = window_data['cc_transaction_all_amt'].sum()
                    cc_cnt = window_data['cc_transaction_all_cnt'].sum()
                    eft_amt = window_data['mobile_eft_all_amt'].sum()
                    eft_cnt = window_data['mobile_eft_all_cnt'].sum()

                    features[f'avg_cc_amt_per_txn_{window}m'] = cc_amt / max(cc_cnt, 1)
                    features[f'avg_eft_amt_per_txn_{window}m'] = eft_amt / max(eft_cnt, 1)
                else:
                    features[f'avg_cc_amt_per_txn_{window}m'] = 0
                    features[f'avg_eft_amt_per_txn_{window}m'] = 0
        else:
            features['months_since_last_txn'] = 999
            features['months_since_last_cc'] = 999
            features['months_since_last_eft'] = 999
            for window in [3, 6, 12]:
                features[f'avg_cc_amt_per_txn_{window}m'] = 0
                features[f'avg_eft_amt_per_txn_{window}m'] = 0

        return features

    def _trend_features(self, cust_history, ref_date):
        """Trend özellikleri"""
        features = {}

        # Son 3 ay vs önceki 3 ay
        recent_3m = ref_date - pd.DateOffset(months=3)
        prev_3m = ref_date - pd.DateOffset(months=6)

        recent_data = cust_history[cust_history['date'] >= recent_3m]
        prev_data = cust_history[(cust_history['date'] >= prev_3m) & (cust_history['date'] < recent_3m)]

        for col in self.txn_cols:
            recent_mean = recent_data[col].mean() if len(recent_data) > 0 else 0
            prev_mean = prev_data[col].mean() if len(prev_data) > 0 else 0

            features[f'{col}_diff_3m'] = recent_mean - prev_mean

            if prev_mean > 0:
                features[f'{col}_pct_chg_3m'] = (recent_mean - prev_mean) / prev_mean
            else:
                features[f'{col}_pct_chg_3m'] = 0

        # Lineer trend (6 ay)
        window_6m = ref_date - pd.DateOffset(months=6)
        trend_data = cust_history[cust_history['date'] >= window_6m].copy()

        if len(trend_data) >= 2:
            trend_data = trend_data.sort_values('date')
            trend_data['time_idx'] = range(len(trend_data))

            for col in self.txn_cols:
                x = trend_data['time_idx'].values
                y = trend_data[col].values

                if len(x) > 1 and y.std() > 0:
                    slope = np.polyfit(x, y, 1)[0]
                    features[f'{col}_slope_6m'] = slope
                else:
                    features[f'{col}_slope_6m'] = 0
        else:
            for col in self.txn_cols:
                features[f'{col}_slope_6m'] = 0

        return features

    def _product_features(self, cust_history, ref_date):
        """Ürün özellikleri"""
        features = {}

        for window in config.WINDOWS:
            window_start = ref_date - pd.DateOffset(months=window)
            window_data = cust_history[cust_history['date'] >= window_start]
            suffix = f"{window}m"

            if len(window_data) > 0:
                features[f'prod_cat_mean_{suffix}'] = window_data['active_product_category_nbr'].mean()
                features[f'prod_cat_max_{suffix}'] = window_data['active_product_category_nbr'].max()
                features[f'prod_cat_min_{suffix}'] = window_data['active_product_category_nbr'].min()
                features[f'prod_cat_last_{suffix}'] = window_data['active_product_category_nbr'].iloc[-1]
            else:
                features[f'prod_cat_mean_{suffix}'] = 0
                features[f'prod_cat_max_{suffix}'] = 0
                features[f'prod_cat_min_{suffix}'] = 0
                features[f'prod_cat_last_{suffix}'] = 0

        # Ürün bırakma
        if len(cust_history) > 0:
            window_6m = ref_date - pd.DateOffset(months=6)
            window_3m = ref_date - pd.DateOffset(months=3)

            data_6m = cust_history[cust_history['date'] >= window_6m]
            data_3m = cust_history[cust_history['date'] >= window_3m]

            if len(data_6m) > 0 and len(data_3m) > 0:
                max_6m = data_6m['active_product_category_nbr'].max()
                last_3m = data_3m['active_product_category_nbr'].mean()

                features['prod_drop_score'] = max_6m - last_3m
                features['is_recent_product_drop'] = int(last_3m < data_6m['active_product_category_nbr'].mean())
            else:
                features['prod_drop_score'] = 0
                features['is_recent_product_drop'] = 0
        else:
            features['prod_drop_score'] = 0
            features['is_recent_product_drop'] = 0

        return features

    def _ratio_features(self, cust_history, ref_date):
        """Oran özellikleri"""
        features = {}

        for window in [3, 6, 12]:
            window_start = ref_date - pd.DateOffset(months=window)
            window_data = cust_history[cust_history['date'] >= window_start]
            suffix = f"{window}m"

            if len(window_data) > 0:
                cc_amt = window_data['cc_transaction_all_amt'].sum()
                cc_cnt = window_data['cc_transaction_all_cnt'].sum()
                eft_amt = window_data['mobile_eft_all_amt'].sum()
                eft_cnt = window_data['mobile_eft_all_cnt'].sum()

                total_amt = cc_amt + eft_amt
                total_cnt = cc_cnt + eft_cnt

                features[f'cc_to_eft_amt_ratio_{suffix}'] = cc_amt / max(eft_amt, 1)
                features[f'cc_to_eft_cnt_ratio_{suffix}'] = cc_cnt / max(eft_cnt, 1)

                # Karışım payları
                features[f'cc_amt_share_{suffix}'] = cc_amt / max(total_amt, 1)
                features[f'eft_amt_share_{suffix}'] = eft_amt / max(total_amt, 1)
                features[f'cc_cnt_share_{suffix}'] = cc_cnt / max(total_cnt, 1)
                features[f'eft_cnt_share_{suffix}'] = eft_cnt / max(total_cnt, 1)
            else:
                features[f'cc_to_eft_amt_ratio_{suffix}'] = 0
                features[f'cc_to_eft_cnt_ratio_{suffix}'] = 0
                features[f'cc_amt_share_{suffix}'] = 0
                features[f'eft_amt_share_{suffix}'] = 0
                features[f'cc_cnt_share_{suffix}'] = 0
                features[f'eft_cnt_share_{suffix}'] = 0

        return features

    def _temporal_features(self, ref_date):
        """Zaman özellikleri"""
        features = {}
        features['ref_month'] = ref_date.month
        features['ref_quarter'] = ref_date.quarter
        features['ref_year'] = ref_date.year

        # Sinüs/kosinüs dönüşümü (mevsimsellik)
        features['ref_month_sin'] = np.sin(2 * np.pi * ref_date.month / 12)
        features['ref_month_cos'] = np.cos(2 * np.pi * ref_date.month / 12)

        return features

    def _demographic_features(self, cust_id):
        """Demografi özellikleri"""
        features = {}

        cust_info = self.customers[self.customers['cust_id'] == cust_id]

        if len(cust_info) > 0:
            cust_info = cust_info.iloc[0]
            features['age'] = cust_info['age'] if pd.notna(cust_info['age']) else -1
            features['tenure'] = cust_info['tenure'] if pd.notna(cust_info['tenure']) else -1
            features['gender'] = cust_info['gender'] if pd.notna(cust_info['gender']) else 'Unknown'
            features['province'] = cust_info['province'] if pd.notna(cust_info['province']) else 'Unknown'
            features['religion'] = cust_info['religion'] if pd.notna(cust_info['religion']) else 'Unknown'
            features['work_type'] = cust_info['work_type'] if pd.notna(cust_info['work_type']) else 'Unknown'
            features['work_sector'] = cust_info['work_sector'] if pd.notna(cust_info['work_sector']) else 'Unknown'
        else:
            features['age'] = -1
            features['tenure'] = -1
            features['gender'] = 'Unknown'
            features['province'] = 'Unknown'
            features['religion'] = 'Unknown'
            features['work_type'] = 'Unknown'
            features['work_sector'] = 'Unknown'

        return features

print("✅ FeatureEngineer sınıfı hazır!")
```

```python
# ═══════════════════════════════════════════════════════════════
# 7. ÖZELLİK ÜRETİMİ
# ═══════════════════════════════════════════════════════════════

# FeatureEngineer oluştur
fe = FeatureEngineer(customer_history, customers)

# Train özellikleri
print("\n🏋️ TRAIN ÖZELLİKLERİ ÜRETİLİYOR...")
X_train_full = fe.create_features(train_df)
y_train = train_df['churn'].values

# Valid özellikleri
print("\n🏋️ VALID ÖZELLİKLERİ ÜRETİLİYOR...")
X_valid_full = fe.create_features(valid_df)
y_valid = valid_df['churn'].values

# Test özellikleri
print("\n🏋️ TEST ÖZELLİKLERİ ÜRETİLİYOR...")
X_test_full = fe.create_features(reference_test)

# Bellek temizliği
gc.collect()

print("\n✅ TÜM ÖZELLİKLER HAZIR!")
print(f"   X_train: {X_train_full.shape}")
print(f"   X_valid: {X_valid_full.shape}")
print(f"   X_test: {X_test_full.shape}")
```

```python
# ═══════════════════════════════════════════════════════════════
# 8. ÖZELLİK İŞLEME
# ═══════════════════════════════════════════════════════════════

# Kategorik sütunları belirle
cat_cols = ['gender', 'province', 'religion', 'work_type', 'work_sector']
num_cols = [col for col in X_train_full.columns if col not in cat_cols + ['cust_id']]

print(f"📊 Sayısal özellik: {len(num_cols)}")
print(f"📊 Kategorik özellik: {len(cat_cols)}")

# Kategorik encoding (Label Encoding)
from sklearn.preprocessing import LabelEncoder

le_dict = {}
for col in cat_cols:
    le = LabelEncoder()

    # Tüm değerleri birleştir (train+valid+test)
    all_values = pd.concat([
        X_train_full[col],
        X_valid_full[col],
        X_test_full[col]
    ]).astype(str).unique()

    le.fit(all_values)

    X_train_full[col] = le.transform(X_train_full[col].astype(str))
    X_valid_full[col] = le.transform(X_valid_full[col].astype(str))
    X_test_full[col] = le.transform(X_test_full[col].astype(str))

    le_dict[col] = le

# cust_id'yi ayır
X_train = X_train_full.drop('cust_id', axis=1)
X_valid = X_valid_full.drop('cust_id', axis=1)
X_test = X_test_full.drop('cust_id', axis=1)

print(f"\n✅ Özellik işleme tamamlandı!")
print(f"   Final shape - Train: {X_train.shape}, Valid: {X_valid.shape}, Test: {X_test.shape}")

# Özellik isimlerini sakla
feature_names = X_train.columns.tolist()
```

```python
# ═══════════════════════════════════════════════════════════════
# 9. MODEL WRAPPER'LARI
# ═══════════════════════════════════════════════════════════════

class ModelWrapper:
    """Tüm modeller için unified interface"""

    def __init__(self, model_type, params, cat_features=None):
        self.model_type = model_type
        self.params = params
        self.cat_features = cat_features
        self.model = None

    def fit(self, X_train, y_train, X_valid=None, y_valid=None):
        """Model eğitimi"""

        if self.model_type == 'lightgbm':
            train_data = lgb.Dataset(X_train, label=y_train)
            valid_data = lgb.Dataset(X_valid, label=y_valid, reference=train_data)

            self.model = lgb.train(
                self.params,
                train_data,
                num_boost_round=config.N_ESTIMATORS,
                valid_sets=[train_data, valid_data],
                valid_names=['train', 'valid'],
                callbacks=[
                    lgb.early_stopping(config.EARLY_STOPPING_ROUNDS),
                    lgb.log_evaluation(0 if not config.VERBOSE else 50)
                ]
            )

        elif self.model_type == 'xgboost':
            dtrain = xgb.DMatrix(X_train, label=y_train)
            dvalid = xgb.DMatrix(X_valid, label=y_valid)

            self.model = xgb.train(
                self.params,
                dtrain,
                num_boost_round=config.N_ESTIMATORS,
                evals=[(dtrain, 'train'), (dvalid, 'valid')],
                early_stopping_rounds=config.EARLY_STOPPING_ROUNDS,
                verbose_eval=0 if not config.VERBOSE else 50
            )

        elif self.model_type == 'catboost':
            train_pool = Pool(X_train, y_train, cat_features=self.cat_features)
            valid_pool = Pool(X_valid, y_valid, cat_features=self.cat_features)

            self.model = CatBoostClassifier(**self.params)
            self.model.fit(
                train_pool,
                eval_set=valid_pool,
                early_stopping_rounds=config.EARLY_STOPPING_ROUNDS,
                verbose=False
            )

        elif self.model_type == 'extratrees':
            self.model = ExtraTreesClassifier(**self.params)
            self.model.fit(X_train, y_train)

        elif self.model_type == 'logistic':
            # Standardize features for LR
            self.scaler = StandardScaler()
            X_train_scaled = self.scaler.fit_transform(X_train)

            self.model = LogisticRegression(**self.params)
            self.model.fit(X_train_scaled, y_train)

        return self

    def predict_proba(self, X):
        """Olasılık tahmini"""

        if self.model_type == 'lightgbm':
            return self.model.predict(X)

        elif self.model_type == 'xgboost':
            dtest = xgb.DMatrix(X)
            return self.model.predict(dtest)

        elif self.model_type == 'catboost':
            return self.model.predict_proba(X)[:, 1]

        elif self.model_type == 'extratrees':
            return self.model.predict_proba(X)[:, 1]

        elif self.model_type == 'logistic':
            X_scaled = self.scaler.transform(X)
            return self.model.predict_proba(X_scaled)[:, 1]

    def get_feature_importance(self, feature_names):
        """Özellik önemini al"""

        if self.model_type == 'lightgbm':
            importance = self.model.feature_importance(importance_type='gain')
            return pd.DataFrame({
                'feature': feature_names,
                'importance': importance
            }).sort_values('importance', ascending=False)

        elif self.model_type == 'xgboost':
            importance = self.model.get_score(importance_type='gain')
            imp_df = pd.DataFrame({
                'feature': list(importance.keys()),
                'importance': list(importance.values())
            }).sort_values('importance', ascending=False)
            return imp_df

        elif self.model_type == 'catboost':
            importance = self.model.get_feature_importance()
            return pd.DataFrame({
                'feature': feature_names,
                'importance': importance
            }).sort_values('importance', ascending=False)

        elif self.model_type in ['extratrees', 'logistic']:
            importance = self.model.feature_importances_ if hasattr(self.model, 'feature_importances_') else np.abs(self.model.coef_[0])
            return pd.DataFrame({
                'feature': feature_names,
                'importance': importance
            }).sort_values('importance', ascending=False)

        return None

print("✅ ModelWrapper hazır!")
```

```python
# ═══════════════════════════════════════════════════════════════
# 10. MODEL EĞİTİMİ
# ═══════════════════════════════════════════════════════════════

models = {}
predictions_valid = {}
predictions_test = {}

# Kategorik özelliklerin indekslerini bul (CatBoost için)
cat_feature_indices = [i for i, col in enumerate(feature_names) if col in cat_cols]

print("\n🚀 MODEL EĞİTİMİ BAŞLIYOR...\n")
print("="*60)

# 1. LightGBM
if config.USE_LGBM:
    print("\n[1/5] 🌳 LightGBM eğitiliyor...")
    lgbm_model = ModelWrapper('lightgbm', config.LGBM_PARAMS)
    lgbm_model.fit(X_train, y_train, X_valid, y_valid)

    predictions_valid['lgbm'] = lgbm_model.predict_proba(X_valid)
    predictions_test['lgbm'] = lgbm_model.predict_proba(X_test)
    models['lgbm'] = lgbm_model

    metrics_lgbm = calculate_all_metrics(y_valid, predictions_valid['lgbm'])
    print_metrics(metrics_lgbm, "LightGBM")

# 2. XGBoost
if config.USE_XGB:
    print("\n[2/5] 🌲 XGBoost eğitiliyor...")
    xgb_model = ModelWrapper('xgboost', config.XGB_PARAMS)
    xgb_model.fit(X_train, y_train, X_valid, y_valid)

    predictions_valid['xgb'] = xgb_model.predict_proba(X_valid)
    predictions_test['xgb'] = xgb_model.predict_proba(X_test)
    models['xgb'] = xgb_model

    metrics_xgb = calculate_all_metrics(y_valid, predictions_valid['xgb'])
    print_metrics(metrics_xgb, "XGBoost")

# 3. CatBoost
if config.USE_CATBOOST:
    print("\n[3/5] 🐱 CatBoost eğitiliyor...")
    cb_model = ModelWrapper('catboost', config.CATBOOST_PARAMS, cat_features=cat_feature_indices)
    cb_model.fit(X_train, y_train, X_valid, y_valid)

    predictions_valid['catboost'] = cb_model.predict_proba(X_valid)
    predictions_test['catboost'] = cb_model.predict_proba(X_test)
    models['catboost'] = cb_model

    metrics_cb = calculate_all_metrics(y_valid, predictions_valid['catboost'])
    print_metrics(metrics_cb, "CatBoost")

# 4. ExtraTrees
if config.USE_EXTRATREES:
    print("\n[4/5] 🌴 ExtraTrees eğitiliyor...")
    et_model = ModelWrapper('extratrees', config.ET_PARAMS)
    et_model.fit(X_train, y_train, X_valid, y_valid)

    predictions_valid['extratrees'] = et_model.predict_proba(X_valid)
    predictions_test['extratrees'] = et_model.predict_proba(X_test)
    models['extratrees'] = et_model

    metrics_et = calculate_all_metrics(y_valid, predictions_valid['extratrees'])
    print_metrics(metrics_et, "ExtraTrees")

# 5. Logistic Regression
if config.USE_LR:
    print("\n[5/5] 📈 Logistic Regression eğitiliyor...")
    lr_model = ModelWrapper('logistic', config.LR_PARAMS)
    lr_model.fit(X_train, y_train, X_valid, y_valid)

    predictions_valid['logistic'] = lr_model.predict_proba(X_valid)
    predictions_test['logistic'] = lr_model.predict_proba(X_test)
    models['logistic'] = lr_model

    metrics_lr = calculate_all_metrics(y_valid, predictions_valid['logistic'])
    print_metrics(metrics_lr, "Logistic Regression")

print("\n" + "="*60)
print("✅ TÜM MODELLER EĞİTİLDİ!")
print("="*60)
```

```python
# ═══════════════════════════════════════════════════════════════
# 11. MODEL KARŞILAŞTIRMA
# ═══════════════════════════════════════════════════════════════

# Tüm model metriklerini topla
all_metrics = {}

for model_name, preds in predictions_valid.items():
    all_metrics[model_name] = calculate_all_metrics(y_valid, preds)

# DataFrame'e çevir
metrics_df = pd.DataFrame(all_metrics).T
metrics_df = metrics_df.sort_values('composite_score', ascending=False)

print("\n📊 MODEL KARŞILAŞTIRMASI (Validation Set)")
print("="*80)
print(metrics_df.to_string())
print("="*80)

# Görselleştirme
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for idx, metric in enumerate(['auc', 'gini', 'recall@10', 'lift@10', 'composite_score']):
    ax = axes[idx]
    metrics_df[metric].plot(kind='barh', ax=ax, color='steelblue')
    ax.set_xlabel(metric.upper())
    ax.set_title(f'{metric.upper()} Comparison')
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f"{config.OUTPUT_DIR}/model_comparison.png", dpi=150, bbox_inches='tight')
plt.show()

print(f"\n💾 Görsel kaydedildi: {config.OUTPUT_DIR}/model_comparison.png")
```

```python
# ═══════════════════════════════════════════════════════════════
# 12. ENSEMBLE & KALİBRASYON
# ═══════════════════════════════════════════════════════════════

print("\n🎯 ENSEMBLE OLUŞTURULUYOR...")

# Her modelin composite score'una göre ağırlık hesapla
weights = {}
total_score = metrics_df['composite_score'].sum()

for model_name in metrics_df.index:
    weights[model_name] = metrics_df.loc[model_name, 'composite_score'] / total_score

print("\n📊 Model Ağırlıkları:")
for model_name, weight in weights.items():
    print(f"   {model_name:15s}: {weight:.4f}")

# Weighted average ensemble
ensemble_valid = np.zeros(len(y_valid))
ensemble_test = np.zeros(len(X_test))

for model_name, weight in weights.items():
    ensemble_valid += weight * predictions_valid[model_name]
    ensemble_test += weight * predictions_test[model_name]

# Ensemble metrikleri
metrics_ensemble_raw = calculate_all_metrics(y_valid, ensemble_valid)
print_metrics(metrics_ensemble_raw, "ENSEMBLE (Raw)")

# Kalibrasyon
if config.CALIBRATION == 'isotonic':
    from sklearn.isotonic import IsotonicRegression

    print("\n🔧 Isotonic Calibration uygulanıyor...")
    calibrator = IsotonicRegression(out_of_bounds='clip')
    calibrator.fit(ensemble_valid, y_valid)

    ensemble_valid_calibrated = calibrator.predict(ensemble_valid)
    ensemble_test_calibrated = calibrator.predict(ensemble_test)

    metrics_ensemble_cal = calculate_all_metrics(y_valid, ensemble_valid_calibrated)
    print_metrics(metrics_ensemble_cal, "ENSEMBLE (Calibrated)")

    # Final predictions
    final_valid_preds = ensemble_valid_calibrated
    final_test_preds = ensemble_test_calibrated

elif config.CALIBRATION == 'platt':
    from sklearn.calibration import calibration_curve
    from sklearn.linear_model import LogisticRegression as PlattScaler

    print("\n🔧 Platt Calibration uygulanıyor...")
    platt = PlattScaler()
    platt.fit(ensemble_valid.reshape(-1, 1), y_valid)

    ensemble_valid_calibrated = platt.predict_proba(ensemble_valid.reshape(-1, 1))[:, 1]
    ensemble_test_calibrated = platt.predict_proba(ensemble_test.reshape(-1, 1))[:, 1]

    metrics_ensemble_cal = calculate_all_metrics(y_valid, ensemble_valid_calibrated)
    print_metrics(metrics_ensemble_cal, "ENSEMBLE (Calibrated)")

    final_valid_preds = ensemble_valid_calibrated
    final_test_preds = ensemble_test_calibrated

else:
    print("\n⚠️  Kalibrasyon uygulanmadı")
    final_valid_preds = ensemble_valid
    final_test_preds = ensemble_test

print("\n✅ ENSEMBLE TAMAMLANDI!")
```

```python
# ═══════════════════════════════════════════════════════════════
# 13. ÖZELLİK ÖNEMİ ANALİZİ
# ═══════════════════════════════════════════════════════════════

print("\n📊 ÖZELLİK ÖNEMİ ANALİZİ...")

# En iyi 3 GBDT modelinden feature importance al
importance_dfs = []

for model_name in ['lgbm', 'xgb', 'catboost']:
    if model_name in models:
        imp_df = models[model_name].get_feature_importance(feature_names)
        imp_df.columns = ['feature', f'{model_name}_importance']
        importance_dfs.append(imp_df)

# Birleştir
if len(importance_dfs) > 0:
    feature_importance = importance_dfs[0]
    for imp_df in importance_dfs[1:]:
        feature_importance = feature_importance.merge(imp_df, on='feature', how='outer')

    # Ortalama importance
    imp_cols = [col for col in feature_importance.columns if 'importance' in col]
    feature_importance['mean_importance'] = feature_importance[imp_cols].mean(axis=1)
    feature_importance = feature_importance.sort_values('mean_importance', ascending=False)

    # Top 20 göster
    print("\n🏆 TOP 20 ÖZELLİK:")
    print("="*60)
    print(feature_importance[['feature', 'mean_importance']].head(20).to_string(index=False))
    print("="*60)

    # Görselleştir
    plt.figure(figsize=(10, 8))
    top_features = feature_importance.head(20)
    plt.barh(range(len(top_features)), top_features['mean_importance'].values, color='coral')
    plt.yticks(range(len(top_features)), top_features['feature'].values)
    plt.xlabel('Mean Importance')
    plt.title('Top 20 Features by Importance')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.savefig(f"{config.OUTPUT_DIR}/feature_importance.png", dpi=150, bbox_inches='tight')
    plt.show()

    print(f"\n💾 Görsel kaydedildi: {config.OUTPUT_DIR}/feature_importance.png")
```

```python
# ═══════════════════════════════════════════════════════════════
# 15. FINAL RAPOR
# ═══════════════════════════════════════════════════════════════

print("\n" + "="*70)
print(" " * 20 + "📊 FINAL RAPOR")
print("="*70)

print("\n🎯 VALIDATION SET PERFORMANSI:")
print("-" * 70)

# En iyi tek model
best_single_model = metrics_df.index[0]
best_single_score = metrics_df.loc[best_single_model, 'composite_score']

print(f"\n🥇 En İyi Tek Model: {best_single_model.upper()}")
print(f"   Composite Score: {best_single_score:.5f}")

# Ensemble performansı
if config.CALIBRATION:
    final_metrics = metrics_ensemble_cal
    print(f"\n🎯 Final Ensemble (Weighted Avg + {config.CALIBRATION.capitalize()} Calibration):")
else:
    final_metrics = metrics_ensemble_raw
    print(f"\n🎯 Final Ensemble (Weighted Avg, No Calibration):")

print(f"   AUC:              {final_metrics['auc']:.5f}")
print(f"   Gini:             {final_metrics['gini']:.5f}")
print(f"   Recall@10%:       {final_metrics['recall@10']:.5f}")
print(f"   Lift@10%:         {final_metrics['lift@10']:.5f}")
print(f"   ───────────────────────────────────")
print(f"   📊 Composite Score: {final_metrics['composite_score']:.5f}")

# Baseline ile karşılaştırma
baseline_composite = 1.0  # Baseline = 1.0 by definition
improvement = ((final_metrics['composite_score'] - baseline_composite) / baseline_composite) * 100

print(f"\n📈 Baseline'a Göre İyileştirme: {improvement:+.2f}%")

# Model ağırlıkları özeti
print("\n⚖️  MODEL AĞIRLIKLARI:")
for model_name, weight in sorted(weights.items(), key=lambda x: x[1], reverse=True):
    print(f"   {model_name:15s}: {weight*100:5.2f}%")

# Top features özet
if len(importance_dfs) > 0:
    print("\n🏆 TOP 10 ÖZELLİKLER:")
    for idx, row in feature_importance.head(10).iterrows():
        print(f"   {idx+1:2d}. {row['feature']:40s} ({row['mean_importance']:.2f})")

print("\n📁 ÇIKTI DOSYALARI:")
print(f"   ✓ {config.OUTPUT_DIR}/submission.csv")
print(f"   ✓ {config.OUTPUT_DIR}/model_comparison.png")
print(f"   ✓ {config.OUTPUT_DIR}/feature_importance.png")

print("\n" + "="*70)
print(" " * 15 + "✅ PROJE TAMAMLANDI!")
print("="*70 + "\n")
```

```python
# ═══════════════════════════════════════════════════════════════
# 17. EK ANALİZLER
# ═══════════════════════════════════════════════════════════════

print("\n📊 EK ANALİZLER...")

# 1. Tahmin dağılımı
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Validation set
axes[0].hist(final_valid_preds, bins=50, alpha=0.7, color='steelblue', edgecolor='black')
axes[0].axvline(final_valid_preds.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {final_valid_preds.mean():.3f}')
axes[0].set_xlabel('Predicted Churn Probability')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Validation Set - Prediction Distribution')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Test set
axes[1].hist(final_test_preds, bins=50, alpha=0.7, color='coral', edgecolor='black')
axes[1].axvline(final_test_preds.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {final_test_preds.mean():.3f}')
axes[1].set_xlabel('Predicted Churn Probability')
axes[1].set_ylabel('Frequency')
axes[1].set_title('Test Set - Prediction Distribution')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f"{config.OUTPUT_DIR}/prediction_distribution.png", dpi=150, bbox_inches='tight')
plt.show()

print(f"   ✓ Tahmin dağılımı grafiği: {config.OUTPUT_DIR}/prediction_distribution.png")

# 2. Calibration curve (validation set)
from sklearn.calibration import calibration_curve

fig, ax = plt.subplots(figsize=(8, 8))

fraction_of_positives, mean_predicted_value = calibration_curve(
    y_valid, final_valid_preds, n_bins=10, strategy='quantile'
)

ax.plot(mean_predicted_value, fraction_of_positives, "s-", label="Model", linewidth=2, markersize=8)
ax.plot([0, 1], [0, 1], "k--", label="Perfect Calibration", linewidth=2)
ax.set_xlabel('Mean Predicted Probability', fontsize=12)
ax.set_ylabel('Fraction of Positives', fontsize=12)
ax.set_title('Calibration Curve (Validation Set)', fontsize=14)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f"{config.OUTPUT_DIR}/calibration_curve.png", dpi=150, bbox_inches='tight')
plt.show()

print(f"   ✓ Kalibrasyon eğrisi: {config.OUTPUT_DIR}/calibration_curve.png")

# 3. Confusion matrix (validation set - optimal threshold)
from sklearn.metrics import confusion_matrix, classification_report

# Optimal threshold (Youden's J statistic)
from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(y_valid, final_valid_preds)
optimal_idx = np.argmax(tpr - fpr)
optimal_threshold = thresholds[optimal_idx]

y_pred_binary = (final_valid_preds >= optimal_threshold).astype(int)

cm = confusion_matrix(y_valid, y_pred_binary)

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax, cbar=True)
ax.set_xlabel('Predicted', fontsize=12)
ax.set_ylabel('Actual', fontsize=12)
ax.set_title(f'Confusion Matrix (Threshold={optimal_threshold:.3f})', fontsize=14)
ax.set_xticklabels(['No Churn', 'Churn'])
ax.set_yticklabels(['No Churn', 'Churn'])

plt.tight_layout()
plt.savefig(f"{config.OUTPUT_DIR}/confusion_matrix.png", dpi=150, bbox_inches='tight')
plt.show()

print(f"   ✓ Confusion matrix: {config.OUTPUT_DIR}/confusion_matrix.png")
print(f"   Optimal threshold: {optimal_threshold:.4f}")

# Classification report
print("\n📋 CLASSIFICATION REPORT (Optimal Threshold):")
print(classification_report(y_valid, y_pred_binary, target_names=['No Churn', 'Churn']))

# 4. Top-K analizi
print("\n📊 TOP-K ANALİZİ (Validation Set):")
print("-" * 60)

for k in [0.05, 0.10, 0.15, 0.20]:
    recall_k = recall_at_k(y_valid, final_valid_preds, k=k)
    lift_k = lift_at_k(y_valid, final_valid_preds, k=k)
    print(f"   Top {int(k*100):2d}%: Recall={recall_k:.4f}, Lift={lift_k:.4f}")

print("\n✅ Ek analizler tamamlandı!")
```
