```python
# Colab'da kameraya erişim yoktur, bu yüzden video yükleyerek veri toplayacağız.
# Çalıştırmadan önce sol panelden 'Dosyalar' sekmesine gidip video dosyanı yükle.


````python
from google.colab import files
uploaded = files.upload()


mp_pose = mp.solutions.pose
pose = mp_pose.Pose()

cap = cv2.VideoCapture(list(uploaded.keys())[0])
data[]
label = "El Salla"

while True:
    ret, frame = cap.read()
    if not ret:
        break

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = pose.process(rgb)

    if results.pose_landmarks:
        landmarks = []
        for lm in results.pose_landmarks.landmark:
            landmarks.extend([lm.x, lm.y, lm.z])
        landmarks.append(label)
        data.append(landmarks)

cap.release()


colums = []

for i in range(33):
  colums.extend([f'x{i}', f'y{i}', f'z{i}'])
colums.append('label')
df = pd.DataFrame(data, columns=colums)
df.to_csv('data.csv', index=False)
df.head()


#Bu hücre ile **hareket verilerini video üzerinden toplar, mediapipe ile landmark çıkarıp etiketleyerek CSV olarak kaydederiz**. Sonraki adımda model eğitime geçeceğiz.
````
